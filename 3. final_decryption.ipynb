```python
from PIL import Image

gif_path = "output_1.gif"
output_folder = "output_frames_1"

gif = Image.open(gif_path)

frame_count = gif.n_frames

# Iterate through each frame and save it as a separate image
for i in range(frame_count):
    gif.seek(i)
    frame = gif.copy()
    frame.save(f"{output_folder}/frame_{i}.png", "PNG")
```

```python
image_1 = Image.open("output_frames_1\\frame_1.png")
image_2 = Image.open("output_frames_2\\frame_3.png")
image_3 = Image.open("output_frames_3\\frame_5.png")
#decryption
share_1 = Image.new("1",(512,512))
share_2 = Image.new("1",(512,512))
share_3 = Image.new("1",(512,512))

for i in range(512):
    for j in range(512):
        pixel_1 = image_1.getpixel((i,j))
        pixel_2 = image_2.getpixel((i,j))
        pixel_3 = image_3.getpixel((i,j))
        pixel_red = pixel_1[0]
        pixel_green = pixel_2[1]
        pixel_blue = pixel_3[2]
        
        binary_red = bin(pixel_red)[2:]
        binary_green = bin(pixel_green)[2:]
        binary_blue = bin(pixel_blue)[2:]
        
        share_1_pixel = int(binary_red[-1])
        share_2_pixel = int(binary_green[-1])
        share_3_pixel = int(binary_blue[-1])
        
        if share_1_pixel == 0:
            share_1.putpixel((i,j),share_1_pixel)
        else:
            share_1.putpixel((i,j),255)
        
        if share_2_pixel == 0:
            share_2.putpixel((i,j),share_2_pixel)
        else:
            share_2.putpixel((i,j),255)
            
        if share_3_pixel == 0:
            share_3.putpixel((i,j),share_3_pixel)
        else:
            share_3.putpixel((i,j),255)
            
            
share_1.save("cameraman/output_share_1.png")
share_2.save("cameraman/output_share_2.png")
share_3.save("cameraman/output_share_3.png")
```

```python
share_1 = Image.open('cameraman/output_share_1.png')
share_2 = Image.open('cameraman/output_share_2.png')
share_3 = Image.open('cameraman/output_share_3.png')
#decrypting the shares into image
width_1, height_1 = share_1.size

# Create a new image for the merged result
decoded_image = Image.new('1', (width_1, height_1))

# Loop through each pixel and overlap the shares
for y in range(height_1):
    for x in range(width_1):
        pixel_value_share1 = share_1.getpixel((x, y))
        pixel_value_share2 = share_2.getpixel((x, y))
        pixel_value_share3 = share_3.getpixel((x, y))
        
        # If both shares have a pixel value of 1, set the merged pixel to 1
        if pixel_value_share1 == 255 and pixel_value_share2 == 255 and pixel_value_share3 == 255:
            decoded_image.putpixel((x, y), 255)
        else:
            decoded_image.putpixel((x, y), 0)

# Save the merged result
decoded_image.save('cameraman/decrypted_image.png')
```
