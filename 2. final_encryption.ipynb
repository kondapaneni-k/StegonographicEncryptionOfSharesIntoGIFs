```python
share_1_path = "cameraman\\3_out_3_share_1.png"
share_2_path = "cameraman\\3_out_3_share_2.png"
share_3_path = "cameraman\\3_out_3_share_3.png"

share_1 = Image.open(share_1_path)
share_2 = Image.open(share_2_path)
share_3 = Image.open(share_3_path)

share_1_pixels = [[0 for j in range(512)] for i in range(512)]
share_2_pixels = [[0 for j in range(512)] for i in range(512)]
share_3_pixels = [[0 for j in range(512)] for i in range(512)]

for i in range(512):
    for j in range(512):
        if share_1.getpixel((i,j)) == 0:
            pass
        else:
            share_1_pixels[i][j] = 1
        if share_2.getpixel((i,j)) == 0:
            pass
        else:
            share_2_pixels[i][j] = 1
        if share_3.getpixel((i,j)) == 0:
            pass
        else:
            share_3_pixels[i][j] = 1
```

```python
# get the secret_coded_frame by embedding shares pixels into selected radom frame pixels at lsb
width, height = frame_1.size

secret_coded_frame_1 = Image.new('RGB',(width,height))
secret_coded_frame_2 = Image.new('RGB',(width,height))
secret_coded_frame_3 = Image.new('RGB',(width,height))


for i in range(width):
    for j in range(height):
        frame_data_1 = frame_1.getpixel((i,j))
        frame_data_2 = frame_2.getpixel((i,j))
        frame_data_3 = frame_3.getpixel((i,j))
        #print(frame_data)

        secret_data_1 = share_1_pixels[i][j]
        secret_data_2 = share_2_pixels[i][j]
        secret_data_3 = share_3_pixels[i][j]
        #print(secret_data_1)

        data_red = frame_data_1[0]
        data_green = frame_data_2[1]
        data_blue = frame_data_3[2]

        binary_representation_red = bin(data_red)[2:]
        binary_representation_green = bin(data_green)[2:]
        binary_representation_blue = bin(data_blue)[2:]
        #print("br",binary_representation_red)

        #insert share_1 data in red part of the pixel
        modified_binary_red = binary_representation_red[:-1] + str(secret_data_1)
        #print("mb1",modified_binary_red)

        #insert share_2 data in green part of the pixel
        modified_binary_green = binary_representation_green[:-1] + str(secret_data_2)

        #insert share_3 data in blue part of the pixel
        modified_binary_blue = binary_representation_blue[:-1] + str(secret_data_3)


        decimal_number_red = int(modified_binary_red, 2)
        decimal_number_green = int(modified_binary_green, 2)
        decimal_number_blue = int(modified_binary_blue, 2)

        secret_coded_frame_1.putpixel((i,j),(decimal_number_red,frame_data_1[1],frame_data_1[2]))
        secret_coded_frame_2.putpixel((i,j),(frame_data_2[0],decimal_number_green,frame_data_2[2]))
        secret_coded_frame_3.putpixel((i,j),(frame_data_3[0],frame_data_3[1],decimal_number_blue))

secret_coded_frame_1.save('frame_1.png')
secret_coded_frame_2.save('frame_3.png')
secret_coded_frame_3.save('frame_5.png')
```

```python
import imageio.v2 as imageio
import numpy as np

def replace_frame(gif_path, frame_path, output_path, frame_number, fps=10):
    # Read the replacement frame
    replacement_frame = imageio.imread(frame_path)

    # Read the original GIF
    gif_reader = imageio.get_reader(gif_path)

    # Create a writer for the modified GIF
    writer = imageio.get_writer(output_path, fps=fps)

    # Iterate through the frames
    for i, frame in enumerate(gif_reader):
        if i == frame_number:
            # Replace the specified frame
            frame = np.array(replacement_frame)

        writer.append_data(frame)

    # Close the writer
    writer.close()

    print(f"Frame {frame_number} replaced successfully. Output saved to {output_path}")

# Example usage
replace_frame('gif1.gif', 'frame_1.png', 'output_1.gif', 1)
```
