```python
from PIL import Image
import numpy as np
from itertools import permutations
import random
```

```python
original_matrix_1 = np.array([[255,255,0,0],[255,0,255,0],[255,0,0,255]])
original_matrix_2 = np.array([[0,0,255,255],[0,255,0,255],[0,255,255,0]])

num_columns_1 = original_matrix_1.shape[1]
num_columns_2 = original_matrix_2.shape[1]

# Get all possible permutations of column indices
column_permutations_1 = permutations(range(num_columns_1))
column_permutations_2 = permutations(range(num_columns_2))


c0 = set()
c1 = set()

# Generate and add matrices with permuted columns to the set
for col_permutation in column_permutations_1:
    permuted_matrix = original_matrix_1[:, col_permutation]
    c0.add(tuple(map(tuple, permuted_matrix)))
for col_permutation in column_permutations_2:
    permuted_matrix = original_matrix_2[:, col_permutation]
    c1.add(tuple(map(tuple, permuted_matrix)))
    
print(len(c0))
print()
print(len(c1))
```

```python
r1 = random.choice(tuple(c0))
r2 = random.choice(tuple(c1))
print(r1[0][0])
print(r2)
```

```python
# opening the image
image_path = 'camera-binary.png'
binary_image = Image.open(image_path)
pixels = list(binary_image.getdata())

width, height = binary_image.size

print(width,height)
print(pixels)
```

```python
share_1 = Image.new('1', (width * 2, height * 2))
share_2 = Image.new('1', (width * 2, height * 2))
share_3 = Image.new('1', (width * 2, height * 2))
#encrypting into shares
# Loop through each pixel in the original image
for y in range(height):
    for x in range(width):
        pixel_value = binary_image.getpixel((x, y))
        print(pixel_value)
        r2 = random.choice(tuple(c0))    # selecting some random matrix to share the white pixels
        r1 = random.choice(tuple(c1))  # selecting some random matrix to sahre the black pixels
        
        # if lena image is being used change pixel_value = (0,0,0) if required output is not coming
        # balck pixels expansion
        if pixel_value== 0:
            share_1.putpixel((2 * x, 2 * y),int(r1[0][0]))
            share_1.putpixel((2 * x + 1, 2 * y),int(r1[0][1]))
            share_1.putpixel((2 * x, 2 * y + 1),int(r1[0][2]))
            share_1.putpixel((2 * x + 1, 2 * y + 1),int(r1[0][3]))
            
            share_2.putpixel((2 * x, 2 * y),int(r1[1][0]))
            share_2.putpixel((2 * x + 1, 2 * y),int(r1[1][1]))
            share_2.putpixel((2 * x, 2 * y + 1),int(r1[1][2]))
            share_2.putpixel((2 * x + 1, 2 * y + 1),int(r1[1][3]))
            
            share_3.putpixel((2 * x, 2 * y),int(r1[2][0]))
            share_3.putpixel((2 * x + 1, 2 * y),int(r1[2][1]))
            share_3.putpixel((2 * x, 2 * y + 1),int(r1[2][2]))
            share_3.putpixel((2 * x + 1, 2 * y + 1),int(r1[2][3]))
        else:
            share_1.putpixel((2 * x, 2 * y),int(r2[0][0]))
            share_1.putpixel((2 * x + 1, 2 * y),int(r2[0][1]))
            share_1.putpixel((2 * x, 2 * y + 1),int(r2[0][2]))
            share_1.putpixel((2 * x + 1, 2 * y + 1),int(r2[0][3]))
            
            share_2.putpixel((2 * x, 2 * y),int(r2[1][0]))
            share_2.putpixel((2 * x + 1, 2 * y),int(r2[1][1]))
            share_2.putpixel((2 * x, 2 * y + 1),int(r2[1][2]))
            share_2.putpixel((2 * x + 1, 2 * y + 1),int(r2[1][3]))
            
            share_3.putpixel((2 * x, 2 * y),int(r2[2][0]))
            share_3.putpixel((2 * x + 1, 2 * y),int(r2[2][1]))
            share_3.putpixel((2 * x, 2 * y + 1),int(r2[2][2]))
            share_3.putpixel((2 * x + 1, 2 * y + 1),int(r2[2][3]))

# Save the new image
share_1.save('cameraman/3_out_3_share_1.png')
share_2.save('cameraman/3_out_3_share_2.png')
share_3.save('cameraman/3_out_3_share_3.png')
```

```python
share_1 = Image.open('cameraman/3_out_3_share_1.png')
share_2 = Image.open('cameraman/3_out_3_share_2.png')
share_3 = Image.open('cameraman/3_out_3_share_3.png')
#decrypting the shares into image
width_1, height_1 = share_1.size

# Create a new image for the merged result
decoded_image = Image.new('1', (width_1, height_1))

# Loop through each pixel and overlap the shares
for y in range(height_1):
    for x in range(width_1):
        pixel_value_share1 = share_1.getpixel((x, y))
        pixel_value_share2 = share_2.getpixel((x, y))
        pixel_value_share3 = share_3.getpixel((x, y))
        
        # If both shares have a pixel value of 1, set the merged pixel to 1
        if pixel_value_share1 == 255 and pixel_value_share2 == 255 and pixel_value_share3 == 255:
            decoded_image.putpixel((x, y), 255)
        else:
            decoded_image.putpixel((x, y), 0)

# Save the merged result
decoded_image.save('cameraman/3_out_3_image.png')
```
