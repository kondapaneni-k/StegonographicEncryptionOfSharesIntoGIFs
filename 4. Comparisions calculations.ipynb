```python
from math import log10, sqrt 
import cv2 
import numpy as np 
from skimage.metrics import structural_similarity as ssim
from skimage import io, img_as_float
```

```python
# psnr for binary image

def PSNR(original, compressed): 
    if original is None or compressed is None:
        raise ValueError("Error: One or both images failed to load.")

    mse = np.mean((original - compressed) ** 2)
    print(mse)
    if mse == 0:  # MSE is zero means no noise is present in the signal. 
                  # Therefore PSNR has no importance. 
        return 100
    max_pixel = 255.0
    psnr = 20 * log10(max_pixel / sqrt(mse)) 
    return psnr 

def main(): 
    original = cv2.imread("3_out_3_image.png", 0)  # Read as grayscale
    compressed = cv2.imread("decrypted_image.png", 0)  # Read as grayscale
    
    try:
        value = PSNR(original, compressed) 
        print(f"PSNR value is {value} dB") 
    except ValueError as e:
        print(e)

if __name__ == "__main__": 
    main()
```

```python
# if for binary image

def calculate_image_fidelity(original_image_path, processed_image_path):
    # Load images
    original_image = img_as_float(io.imread(original_image_path, as_gray=True))
    processed_image = img_as_float(io.imread(processed_image_path, as_gray=True))

    # Calculate SSIM
    ssim_index, _ = ssim(original_image, processed_image, full=True)

    # Return the SSIM index
    return ssim_index

# Example usage
original_image_path = '3_out_3_image.png'
processed_image_path = 'decrypted_image.png'

fidelity_score = calculate_image_fidelity(original_image_path, processed_image_path)
print(f"Structural Similarity Index (SSI): {fidelity_score}")
```

```python
# accuracy

def calculate_pixel_accuracy(original_image, generated_shares):
    # Load the original binary image and generated shares
    original_image = cv2.imread(original_image, cv2.IMREAD_GRAYSCALE)
    generated_shares = cv2.imread(generated_shares, cv2.IMREAD_GRAYSCALE)

    # Ensure the images have the same dimensions
    if original_image.shape != generated_shares.shape:
        raise ValueError("Images must have the same dimensions.")

    # Calculate pixel-wise accuracy
    total_pixels = original_image.size
    correct_pixels = np.sum(original_image == generated_shares)
    accuracy = (correct_pixels / total_pixels) * 100

    return accuracy

# Example usage
original_image_path = '3_out_3_image.png'
generated_shares_path = 'decrypted_image.png'

accuracy = calculate_pixel_accuracy(original_image_path, generated_shares_path)

print(f"Pixel-wise Accuracy: {accuracy:.2f}%")
```
